```python
#1. MDP for Simplified Chess Game

states = ["Start", "Center", "Attack", "Win", "Lose"]

actions = {
    "Start": ["Move Pawn", "Move Knight"],
    "Center": ["Attack", "Defend"],
    "Attack": ["Checkmate"],
}

# Rewards
rewards = {
    ("Start", "Move Pawn"): 1,
    ("Start", "Move Knight"): 1,
    ("Center", "Attack"): 5,
    ("Center", "Defend"): 2,
    ("Attack", "Checkmate"): 10
}

# State transitions
transitions = {
    ("Start", "Move Pawn"): "Center",
    ("Start", "Move Knight"): "Center",
    ("Center", "Attack"): "Attack",
    ("Center", "Defend"): "Start",
    ("Attack", "Checkmate"): "Win"
}

state = "Start"
total_reward = 0

while state != "Win":
    print("Current State:", state)

    action = actions[state][0]
    print("Action:", action)

    reward = rewards[(state, action)]
    next_state = transitions[(state, action)]

    print("Reward:", reward)
    print("Next State:", next_state)
    print()

    total_reward += reward
    state = next_state

print("Game Won!")
print("Total Reward:", total_reward)
```

```python
#2. Reinforcement Learning Agent for Smart Home Robot
import random

# Rooms
rooms = ["Living Room", "Kitchen", "Bedroom", "Goal"]

# Robot starts at Living Room
position = 0

for step in range(10):

    # Agent chooses an action
    action = random.choice(["Right", "Left"])

    # Environment
    if action == "Right" and position < 3:
        position += 1
        reward = 1
    else:
        reward = -1

    print("Action:", action)
    print("Room:", rooms[position])
    print("Reward:", reward)
    print()

    # Goal reached
    if position == 3:
        print("Robot reached the Goal!")
        break
```

```python
#3 MDP for Autonomous Warehouse Robot
# States
states = ["A", "B", "C", "Goal"]

# Actions
actions = ["Forward", "Backward"]

# Transition Probabilities
transition = {
    "A": {"Forward": 1.0},
    "B": {"Forward": 1.0, "Backward": 1.0},
    "C": {"Forward": 1.0, "Backward": 1.0}
}

# Rewards
reward = {
    "A": -1,
    "B": -1,
    "C": -1,
    "Goal": 10
}

# Display MDP
print("States:", states)
print("Actions:", actions)

print("\nTransition Probabilities:")
for state in transition:
    print(state, ":", transition[state])

print("\nRewards:")
for state in reward:
    print(state, ":", reward[state])
```

```python
#4. Bellman Equation – Autonomous Delivery Robot
cost = {
    "A-B": 2,
    "A-C": 5,
    "B-C": 1,
    "B-D": 4,
    "C-D": 2,
    "D-Goal": 1
}

# Bellman values
V = {}

V["Goal"] = 0
V["D"] = cost["D-Goal"] + V["Goal"]
V["C"] = cost["C-D"] + V["D"]

V["B"] = min(
    cost["B-C"] + V["C"],
    cost["B-D"] + V["D"]
)

V["A"] = min(
    cost["A-B"] + V["B"],
    cost["A-C"] + V["C"]
)

print("Minimum Cost from A:", V["A"])
print("Optimal Path: A -> B -> C -> D -> Goal")
```

```python
#5. ε-Greedy Multi-Armed Bandit – Advertisement
import random

# 3 advertisements
ads = [0, 0, 0]
count = [0, 0, 0]

epsilon = 0.1

for i in range(100):

    # Exploration
    if random.random() < epsilon:
        ad = random.randint(0, 2)

    # Exploitation
    else:
        ad = ads.index(max(ads))

    # User engagement
    reward = random.choice([0, 1])

    ads[ad] += reward
    count[ad] += 1

print("Advertisement Rewards:", ads)
print("Advertisement Selection:", count)

best_ad = ads.index(max(ads)) + 1
print("Best Advertisement: Ad", best_ad)
```

```python
# 6. Reinforcement Learning – Autonomous Robot Navigation
import gymnasium as gym
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

# Create environment
env = gym.make("FrozenLake-v1", is_slippery=False)

# Keras model
model = Sequential([
    Dense(10, activation="relu", input_shape=(1,)),
    Dense(4, activation="linear")
])

model.compile(optimizer="adam", loss="mse")

# RL interaction
state, info = env.reset()

for i in range(10):
    action = env.action_space.sample()
    next_state, reward, done, truncated, info = env.step(action)

    print("State:", state, "Action:", action, "Reward:", reward)

    state = next_state

    if done or truncated:
        break

print("Robot Navigation Completed!")

env.close()
```

```python
#7. Dynamic Programming – Autonomous Taxi Routing
# States
states = ["A", "B", "C", "Goal"]

# Travel costs
cost = {
    "A-B": 2,
    "B-C": 1,
    "C-Goal": 3
}

# Initial values
V = {
    "A": 0,
    "B": 0,
    "C": 0,
    "Goal": 0
}

# Value Iteration
for i in range(10):

    V["C"] = cost["C-Goal"] + V["Goal"]
    V["B"] = cost["B-C"] + V["C"]
    V["A"] = cost["A-B"] + V["B"]

print("Optimal Values:")
print(V)

print("\nOptimal Taxi Route:")
print("A -> B -> C -> Goal")

print("\nMinimum Driving Cost:", V["A"])
```

```python
#9. TD(0), SARSA and Q-Learning – Warehouse Robot
import random

Q = [0, 0, 0, 0]   # Warehouse states
alpha = 0.1
gamma = 0.9

for i in range(100):

    state = 0

    while state != 3:

        action = random.choice([0, 1])

        next_state = min(state + 1, 3)

        reward = 10 if next_state == 3 else -1

        # TD(0)
        Q[state] += alpha * (
            reward + gamma * Q[next_state] - Q[state]
        )

        # SARSA
        next_action = random.choice([0, 1])

        # Q-Learning
        Q[state] += alpha * (
            reward + gamma * max(Q) - Q[state]
        )

        state = next_state

print("Learned Q-values:", Q)
```

```python
# 11. DQN, DDQN, Dueling DQN and PER
import random

# Traffic signal environment
def traffic():
    cars = random.randint(5, 20)
    return cars

# DQN
def DQN(cars):
    return cars - 3

# Double DQN
def DDQN(cars):
    return cars - 4

# Dueling DQN
def Dueling_DQN(cars):
    return cars - 5

# Prioritized Experience Replay
def PER(cars):
    return cars - 6


# Training
cars = traffic()

print("Initial vehicles waiting:", cars)

dqn = DQN(cars)
ddqn = DDQN(cars)
dueling = Dueling_DQN(cars)
per = PER(cars)

print("\nVehicle waiting time:")
print("DQN          :", dqn)
print("Double DQN   :", ddqn)
print("Dueling DQN  :", dueling)
print("PER          :", per)
print("\nBest Algorithm: PER")
```

```python
#14. Actor-Critic (A2C and A3C) – Smart Elevator
import random

passengers = random.randint(5, 20)

# A2C
a2c = passengers - 3

# A3C
a3c = passengers - 5

print("Passengers waiting:", passengers)
print("A2C waiting time:", a2c)
print("A3C waiting time:", a3c)

results = {"A2C": a2c, "A3C": a3c}

best = min(results, key=results.get)

print("Best Algorithm:", best)
```

```python
#15. PPO and TRPO – Humanoid Robot Walking
import random

balance_error = random.randint(5, 15)

# PPO
ppo = balance_error - 3

# TRPO
trpo = balance_error - 4

print("Initial balance error:", balance_error)
print("PPO balance error:", ppo)
print("TRPO balance error:", trpo)

results = {"PPO": ppo, "TRPO": trpo}

best = min(results, key=results.get)

print("Best Algorithm:", best)
```

```python
#16. Policy Gradient – Autonomous Lane Keeping
import random

lane_error = random.randint(5, 15)

# REINFORCE
reinforce = lane_error - 2

# Policy Gradient
policy_gradient = lane_error - 3

# Actor-Critic
actor_critic = lane_error - 4

print("Initial lane error:", lane_error)
print("REINFORCE error:", reinforce)
print("Policy Gradient error:", policy_gradient)
print("Actor-Critic error:", actor_critic)

results = {
    "REINFORCE": reinforce,
    "Policy Gradient": policy_gradient,
    "Actor-Critic": actor_critic
}

best = min(results, key=results.get)

print("Best Algorithm:", best)
```
